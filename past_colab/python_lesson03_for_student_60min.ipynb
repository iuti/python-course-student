## 15〜27分｜`range()`を調べよう

`i`には、繰り返すたびに`range()`から数字が1つずつ入る。

実行前に表示される数字を予想しよう。

```python
for i in range(5):
    print(i)
```

`range(5)`は、0から始まり、5の直前で止まる。

では、次のコードは何を表示する？

```python
for shot in range(1, 6):
    print(shot, "発目！")
```

## 27〜38分｜ダメージを合計しよう

`total_damage`に、1発ごとのダメージを足していく。

実行前に、最後の合計ダメージを予想しよう。

```python
total_damage = 0

for shot in range(1, 6):
    total_damage = total_damage + 20
    print(shot, "発目：合計", total_damage, "ダメージ")

print("最終ダメージ:", total_damage)
```
