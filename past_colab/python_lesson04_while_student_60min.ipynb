## 24〜36分｜敵を倒すまで攻撃

敵のHPが0より大きい間、30ダメージの攻撃を繰り返す。

実行前に考えよう。

- 攻撃は何回行われる？
- HPはどの順番で変化する？
- 最後のHPはいくつ？

```python
enemy_hp = 100

while enemy_hp > 0:
    enemy_hp = enemy_hp - 30
    print("攻撃！ 残りHP:", enemy_hp)

print("敵を撃破！")
```

### 攻撃回数も数えよう

`turn`を0から始め、攻撃のたびに1を足す。

```python
enemy_hp = 100
damage = 30
turn = 0

while enemy_hp > 0:
    enemy_hp = enemy_hp - damage
    turn = turn + 1
    print(turn, "回目の攻撃！ 残りHP:", enemy_hp)

print(turn, "回で敵を撃破！")
```
