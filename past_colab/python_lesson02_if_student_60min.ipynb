## 30〜40分｜ランク判定を作ろう

条件が3種類以上あるときは`elif`を使える。

実行前に **59・60・79・80** の結果を予想しよう。

```python
score = int(input("スコア: "))

if score >= 80:
    print("Aランク")
elif score >= 60:
    print("Bランク")
else:
    print("Cランク")
```
